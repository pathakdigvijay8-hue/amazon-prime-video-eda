####Handling Missing Values


````python
titles_missing = titles.isnull().sum() # Titles Missing Values
display(titles_missing)
````

````python
# Cleaning Titles Dataset
````

````python
# Movies don't have seasons, so 0 is factually correct
titles['seasons'] = titles['seasons'].fillna(0)
````

````python
# filling categorical missing values
titles['description'] = titles['description'].fillna('No Description')
titles['age_certification'] = titles['age_certification'].fillna('Unknown')
````

````python
# filling numerical missing values
titles['imdb_score'] = titles['imdb_score'].fillna(titles['imdb_score'].mean())
titles['tmdb_score'] = titles['tmdb_score'].fillna(titles['tmdb_score'].median())
titles['tmdb_popularity'] = titles['tmdb_popularity'].fillna(titles['tmdb_popularity'].median())
titles['imdb_votes'] = titles['imdb_votes'].fillna(titles['imdb_votes'].median())
````

````python
# Cleaning Credits Dataset
````

````python
credits['character'] = credits['character'].fillna('Unknown')
````

````python
titles_missing = titles.isnull().sum()
display(titles_missing)
````

#### Handling Duplicates

````python
print("Initial exact duplicates in Titles:", titles.duplicated().sum())
print("Initial exact duplicates in Credits:", credits.duplicated().sum())
````

````python
# Deleting Duplicates
titles = titles.drop_duplicates(subset=['id'], keep='first')
credits = credits.drop_duplicates(subset=['id', 'person_id', 'role'], keep='first')
````

####Merging Datasets

````python
# Merging credits and titles datasets using left merge
df_merged = pd.merge(titles, credits, on='id', how='left')
````

### What all manipulations have you done and insights you found?

### Data Manipulations I Performed

* **Handled Missing Seasons:** I replaced missing values in the `seasons` column with `0`, since these empty rows represented movies, which naturally do not have seasons.
* **Filled Categorical Data:** Instead of deleting rows with missing text, I filled the `description`, `age_certification`, and `character` columns with standard placeholders like 'No Description' and 'Unknown'.
* **Imputed Numerical Data:** For missing numerical ratings, I filled the blanks using the mean for `imdb_score`, and the median for `tmdb_score`, `tmdb_popularity`, and `imdb_votes`.
* **Removed Exact Duplicates:** I removed duplicates by targeting specific identifiers. I filtered the titles using just the `id`, and filtered the credits using a combination of `id`, `person_id`, and `role`, making sure to keep only the first occurrence.
* **Merged Datasets:** I combined the `titles` and `credits` tables into a single dataset by joining them on their shared `id` column.

---

### Key Insights from the Data

* **Identifying Content Types:** By fixing the `seasons` column (filling blanks with 0), it became easy to mathematically distinguish between movies (0 seasons) and TV shows (1 or more seasons) for future analysis.
* **Handling Skewed Data:** When looking at popularity and vote counts, I realized a few blockbuster titles have massively higher numbers than the rest. Using the median to fill missing values protected my dataset from being artificially inflated by those extreme outliers.
* **Understanding Cast Roles:** When checking for duplicates, I realized that a single person could have multiple roles in the same movie (like acting and directing). Filtering duplicates by `person_id` *and* `role` ensured I didn't accidentally delete these valid, multi-talented records.
* **Preserving the Dataset Size:** By choosing to fill missing values with placeholders (like 'Unknown') instead of deleting those rows, I successfully retained thousands of valid data points (like actors and movie titles) that would have otherwise been lost.
